```python
def max_element(lst):
    if not lst:
        return None
        
    max_val = lst[0]
    for num in lst:
        if num > max_val:
            max_val=num
    return max_val

max_element([])
```

```python
def second_largest(nums):
    largest = float('-inf')
    second = float('-inf')

    for num in nums:
        if num > largest:
            second = largest
            largest = num

        elif num > second and num != largest:
            second = num
```

```python
def sum_list(nums):
    summ = 0
    for num in nums:
        summ+=num

    return summ
```

```python
def palindrome_list(nums):
    start = 0
    end = len(nums) - 1
    while start<end:
        if nums[start] != nums[end]:
            return False
        start+=1
        end-=1
    return True
```

```python
def reverse_list(nums):
    start = 0
    end = len(nums) - 1
    while start<end:
        nums[start],nums[end]=nums[end],nums[start]
        start+=1
        end-=1
    return nums
```

```python
def rotate_list(nums,d):
    new = []
    for i in range(d,len(nums)):
        new.append(nums[i])
    for i in range(0,d):
        new.append(nums[i])

    return new
```
