```python
def generate_square(n):
    square = []

    for i in range(n):
        row='*'*n
        square.append(row)

    return square

print(generate_square(3))
```

```python
def hollow_square(n):
    lst=[]
    for i in range(n):
        if i==0 or i==n-1:
            lst.append('*'*n)
        else:
            row='*'+' '*(n-2) + '*'
            lst.append(row)

    return lst

print(hollow_square(5))
            
```

```python
def rectangle(n,m):
    lst=[]
    for i in range(n):
        lst.append('*'*m)
    return lst

rectangle(4,5)
```

```python
def right_angle_triangle(n):
    lst=[]
    for i in range(1,n+1):
        lst.append("*"*i)
    return lst

right_angle_triangle(5)
```
