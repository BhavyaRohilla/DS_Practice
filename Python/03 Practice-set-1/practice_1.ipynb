```python
def generate_square(n):
    square = []

    for i in range(n):
        row='*'*n
        square.append(row)

    return square

print(generate_square(3))
```

```python
def hollow_square(n):
    lst=[]
    for i in range(n):
        if i==0 or i==n-1:
            lst.append('*'*n)
        else:
            row='*'+' '*(n-2) + '*'
            lst.append(row)

    return lst

print(hollow_square(5))
            
```

```python
def rectangle(n,m):
    lst=[]
    for i in range(n):
        lst.append('*'*m)
    return lst

rectangle(4,5)
```

```python
def right_angle_triangle(n):
    lst=[]
    for i in range(1,n+1):
        lst.append("*"*i)
    return lst

right_angle_triangle(5)
```

```python
def inverted_right_triangle(n):
    lst = ['*'*i for i in range(n,0,-1)]
    return lst

inverted_right_triangle(3)
```

```python
def pyramid_pattern(n):
    return [' ' * (n-i-1) + '*'*(2*i+1) for i in range(n)]
l1 = pyramid_pattern(3)

for i in l1:
    print(i)
```

```python
def inverted_pyramid_pattern(n):
    return [' ' * i + '*'*(2*(n-i)-1) for i in range(n)]
l1 = inverted_pyramid_pattern(4)
for i in l1:
    print(i)
```

```python
def right_angle_triangle(n):
    return [str(i)*i for i in range(1,n+1)]

right_angle_triangle(5)
```

```python
def floyd_triangle(n):
    num = 1
    result = []

    for i in range(1,n+1):
        row=' '.join(str(num+j) for j in range(i))
        result.append(row)
        num+=1
    return result

floyd_triangle(4)
```

```python
def daimond(n):
    upper = [' ' * (n-i-1) + '*' * (2*i+1) for i in range(n)]
    lower = [' ' * i + '*' * (2*(n-i-1)+1) for i in range(n)]
    return upper+lower

daimond(5)
```
