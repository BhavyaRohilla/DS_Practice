##### Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs.
Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## Creating Dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student = {"name": "Bhavya", "age": 25, "grade":24}
print(student)
print(type(student))
```

```python
# Single key is always used
student = {"name": "Bhavya", "age": 25, "name":24}
print(student)
```

```python
## Accessing Dictionary Element
student = {"name": "Bhavya", "age": 25, "grade":24}
print(student)
```

```python
## Accessing Dictionary elements
print(student['grade'])
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not Available"))
```

```python
## Modifying dictionary Element
## Dictionary are mutable, So you can add, update or delete element
print(student)
```

```python
student["age"]=33 ## Update value for the key
print(student)
student["address"] = "india" ## Added a new key and value
print(student)
```

```python
del student['grade'] ## Delete key and value pair
print(student)
```

```python
## Dictionary methods
keys=student.keys() # get all the keys
print(keys)
values=student.values() # get all the values
print(values)

items=student.items() # get all key value pairs
print(items)
```

```python
## Shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="mahak"
print(student_copy)
print(student)
```

```python
student_copy1 = student.copy() ## Shallow copy
print(student_copy1)
print(student)
```

```python
student["name"]="Radha"
student_copy1["age"]=55
print(student)
print(student_copy1)
print(student_copy)
```

```python
###iterating over dictionary
## YOu can use loops to iterate over dictionary

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for value in student.values():
    print(value)
```

```python
## iterate over key value pairs
for key,value in student.items():
    print(f"{key} : {value}")
```

```python
## Nested dictionary
students={
    "student1": {"name": "Bhavya", "age": 25},
    "student2": {"name": "Radha", "age": 22}
}

print(students)
```

```python
## Access nested dictionary elements
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary comprehension
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
## condition Dictionary comprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical examples

## Use a dictrionary to count the frequncy of element in the list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge 2 dictionaries into one
dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```

#### Conclusion
Dictionaries are powerful tools in Python for managing key-value pairs. They are used in a variety of real-world scenarios, such as counting word frequency, grouping data, storing configuration settings, managing phonebooks, tracking inventory, and caching results.
