##### Introduction to Functions
Definition:

A function is a block of code that performs a specific task.
Functions help in organizing code, reusing code, and improving readability.



```python
## Syntax
def function_name(parameters):
    """Docstring"""
    # Function body
    return expression
```

```python
## why functions?
num = 24
if num%2==0:
    print("The number is even")
else:
    print("The number is odd")
    
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
```

```python
## Call the function
even_or_odd(24)
```

```python
## function with multiple parameters

def add(a,b):
    return a+b

result=add(2,4)
print(result)
```

```python
## Default parameters
def greet(name="guest"):
    print(f"Hello {name} Welcome to the paradise")

greet()
```

```python
### Variable Length Argement
## Positional and Keyboard argement

def print_numbers(*bhavya):
    for number in bhavya:
        print(number)

print_numbers(1,2,3,4,5,6,7,8,"Bhavya")
```

```python
## Positional argument
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
## Keyword Argument

def print_details(**k):
    for key,value in k.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Bhavya",age="25",country="India")
```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional argement :{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
    
```

```python
print_details(1,2,3,4,"Bhavya", name="Bhavya",age="32",country="india")
```

```python
## Return statement
def multiply(a,b):
    return a*b

multiply(2,3)
```

```python
### Return multiple parameters
def multiply(a,b):
    return a*b,a

multiply(2,3)
```
