```python
def count_negative(grid):
    count = 0
    n = len(grid[0])

    for row in grid:
        left,right = 0,n-1
        first_negative = n

        while left <= right:
            mid = (left+right)//2
            if row[mid]<0:
                first_negative = mid
                right = mid - 1
            else:
                left = mid + 1

        count = count + n - first_negative
    return count

grid = [
    [4, 3, 2, 1],
    [3, 2, 1, -1],
    [1, 1, -1, -2],
    [-1, -1, -2, -3]
]
print("length",grid[0])
for row in grid:
    print(row)
result = count_negative(grid)

print(result)
```

```python
def smallest_leter(words,target):
    start = 0
    end = len(words)-1
    ans = 0
    while start <= end:
        mid = (start+end)//2
        if(ord(words[mid]) > ord(target)):
            ans = mid
            end = mid-1
        else:
            start = mid+1

    return words[ans]

words = ['c', 'f', 'j']
        
print(smallest_leter(words, 'k'))  # c
print(smallest_leter(words, 'c'))  # f
print(smallest_leter(words, 'a'))  # c

```

```python
def first_last_pos(lst,target):
    start = 0
    end = len(lst)-1
    st_pos = -1

    while(start <= end):
        mid = (start+end)//2
        if lst[mid]==target:
            st_pos = mid
            end = mid - 1
        elif lst[mid] > target:
            end = mid - 1
        else:
            start = mid + 1

    start = 0
    end = len(lst)-1
    en_pos = -1
    
    while(start <= end):
            mid = (start+end)//2
            if lst[mid]==target:
                en_pos = mid
                start = mid+ 1
            elif lst[mid] > target:
                end = mid - 1
            else:
                start = mid + 1

    return [st_pos, en_pos]

print(first_last_pos([5, 7, 7, 8, 8, 10], 8))
print(first_last_pos([5, 7, 7, 8, 8, 10], 6))
```

```python
def minimum(lst):
    start = 0
    end = len(lst) - 1
    while start<=end:
        mid = (start + end)//2
        if lst[mid] > lst[end]:
            start = mid+1
        else:
            end = mid

    return lst[start]  

print(minimum([4, 5, 6, 7, 0, 1, 2]))  # 0
print(minimum([11, 13, 15, 17]))     
```
