```python
def largest(lst):
    temp = lst[0]
    for i in range(1,len(lst)):
        if lst[i]>temp:
            temp = lst[i]

    return temp

```

```python
def remove_deplicate(lst):
    result = []
    for num in lst:
        if num not in result:
            result.append(num)
    return result

print(remove_deplicate([1,2,2,3,3,4,4,5,5]))
```

```python
def remove_duplicates(lst):
    seen = {}
    for num in lst:
        if num in seen:
            return False

        seen[num] = True

    return True
```

```python
def reverse(lst):
    start = 0
    end = len(lst)-1
    while(start<end):
        lst[start],lst[end] = lst[end],lst[start]
        start+=1
        end-=1

    return lst
```

```python
def count_even_odd(lst):
    even_count = 0
    odd_count = 0
    for i in lst:
        if i%2==0:
            even_count += 1
        else:
            odd_count += 1
      

    return even_count,odd_count
```

```python
def max_diff(lst):
    maximum = 0
    for i in range(len(lst)-1):
        difference = abs(lst[i] - lst[i+1])
        if difference > maximum:
            maximum = difference
    return maximum

max_diff([1,7,3,10,5])
```

```python
def merge_two_list(a,b):
    i = j = 0
    result = []

    while i<len(a) and j<len(b):
        if a[i] <=b[j]:
            result.append(a[i])
            i+=1
        else:
            result.append(b[j])
            j+=1
    result.extend(a[i:])
    result.extend(b[j:])
    return result


print(merge_two_list([1,2,3,4],[4,5,6,7]))
```

```python
def rotate_list(lst,k):
    result = []
    for i in range(-k,0):
        result.append(lst[i])
    for i in range(0,len(lst)-k):
        result.append(lst[i])

    return result

rotate_list([1,2,3,4,5],2)
```

```python
def merge_lists_to_dictionary(keys, values):
    if len(keys) != len(values):
        return False

    result = {}

    for i in range(len(keys)):
        result[keys[i]] = values[i]

    return result
```

```python
def merge_three_dict(dict1,dict2,dict3):
    result={}
    for i in (dict1,dict2,dict3):
        for key in i:
            result[key] = i[key]

    return result

dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3}
dict3 = {'a': 10, 'd': 4}
merge_three_dict(dict1,dict2,dict3)
```
