```python
def reverse_string(str):
    result = list(str)
    start = 0
    end =len(str)-1
    while start<end:
        result[start],result[end]=result[end],result[start]
        start+=1
        end-=1

    return ''.join(result)

reverse_string("Bhavya")
```

```python
def count_vowels(s):
    count = 0
    for i in s:
        if i in "aeiouAEIOU":
            count+=1
    return count
```

```python
def are_equal_strings(s1,s2):
    if len(s1) != len(s2):
        return False
    for i in range(len(s1)):
        if s1[i] != s2[i]:
            return False
    return True
```

```python
def is_palindrome1(s):
    left = 0
    right = len(s)-1

    while left < right:

        if not s[left].isalnum():
            left+=1
            continue

        if not s[right].isalnum():
            right-=1
            continue

        if s[left].lower() != s[right].lower():
            return False

        left+=1
        right-=1
    return True

print(is_palindrome1("A man a plan a canal Panama"))
```

```python
def count_words(s):
    words = s.split()
    return len(words)
count_words("Hello, world")
```

```python
def count_words1(s):
    count = 0
    in_word = False

    for char in s:
        if char != " " and not in_word:
            count+=1
            in_word=True

        elif char == " ":
            in_word = False

    return count
```

```python
def remove_duplicate(s):
    result=""
    for ch in s:
        if ch not in result:
            result +=ch
    return result


remove_duplicate("programming")
```

```python
def remove_duplicate1(s):
    seen = set()
    result = []

    for ch in s:
        if ch not in seen:
            result.append(ch)
            seen.add(ch)
    return "".join(result)

remove_duplicate("programming")
```

```python
def count_consonent(s):
    count = 0
    for i in s:
        if i.isalpha() and i not in "aeiouAEIOU":
            count+=1
    return count
```

```python
def check_anagram(s,t):
    if len(s) != len(t):
        return False
    dict1 = {}
    dict2 = {}
    for i in range(len(s)):
        dict1[s[i]] = dict1.get(s[i], 0)+1
    for i in range(len(t)):
        dict2[t[i]] = dict2.get(t[i], 0)+1

    return dict1 == dict2

    
```

```python
def check_anagram1(s,t):
    if len(s) != len(t):
        return False
    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in t:
        if ch not in freq:
            return False

        freq[ch] -= 1

        if freq[ch] < 0:
            return False
    return True

check_anagram1("anagram","nagaram")
```

```python
def check_subsequence(s, t):
    j = 0

    for i in range(len(s)):
        if s[i] == t[j]:
            j += 1

        if j == len(t):
            return True

    return False
```
