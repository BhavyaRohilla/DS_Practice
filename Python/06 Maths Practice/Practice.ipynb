```python
def even_sum(n):
    for i in range(1,n+1):
        vl = 2*i
        even_sum = 0
        even_sum += vl
    return even_sum 
```

```python
def even_sum1(n):
    return n*(n+1)
```

```python
def check_even(num):
    return num%2==0
```

```python
def prime_number(num):
    if num < 2:
        return False
    for i in range(2,num//2+1):
        if num%i==0:
            return False

    return True
```

```python
def prime_number(num):
    if num < 2:
        return False
    for i in range(2,int(num**0.5)+1):
        if num%i==0:
            return False

    return True
```

```python
def perfect_square(num):
    for i in range(num//2+1):
        if i*i == num:
            return True
        else
            return False
```

```python
def is_perfect_square(num):
    if num < 1:
        return False

    left= 1
    right = num

    while left<=right:
        mid = (left+right)//2
        square = mid*mid

        if square == num:
            return True
        elif square < num:
            left = mid + 1
        else:
             right = mid - 1
    return False
```

```python
def int_to_binary(n):
    if n == 0:
        return "0"

    sign = ""

    if n<0:
        sign = "-"
        n = abs(n)

    result = []

    while n>0:
        remainder = n%2
        result.append(str(remainder))
        n = n//2

    return sign + "".join(result[::-1])

int_to_binary(25)
```

```python
def int_to_binary1(n):
    if n==0:
        return "0"

    sign=""
    if n<0:
        sign="-"
        n=abs(n)

    result = 0
    place = 1

    while n>0:
        remainder = n%2
        result += remainder*place
        place = place*10
        n=n//2

    return sign + str(result)

int_to_binary1(25)
```
